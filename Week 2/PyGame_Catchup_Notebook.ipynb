## 🔄 Section 2 — The Game Loop
Every game ever made — from Pong to Minecraft — follows the same basic structure. Think of it as a recipe with three steps:

```
┌─────────────────────────────────┐
│  1. INITIALISE                  │  Runs ONCE
│     Set up window, variables,   │  (like setting up a board game)
│     images, sounds              │
└──────────────┬──────────────────┘
               │
               ▼
┌─────────────────────────────────┐
│  2. GAME LOOP  ◄────────────┐   │  Runs ~60× per second
│                             │   │
│   a) Handle Events          │   │  Events  = keyboard, mouse, quit
│   b) Update State      ─────┘   │  Update  = move things, check score
│   c) Draw Frame                 │  Draw    = clear screen, draw, show
└──────────────┬──────────────────┘
               │  (quit event received)
               ▼
┌─────────────────────────────────┐
│  3. QUIT                        │  Runs ONCE
│     pygame.quit() + sys.exit()  │
└─────────────────────────────────┘
```

Each pass through the loop is called a **frame**. At 60 FPS (frames per second), the loop runs 60 times every second — which is why games look smooth.


### 🏷️ Label the Game Loop
Read the code below. In the comment after each section, write which part it is: `INITIALISE`, `GAME LOOP`, or `QUIT`.


````python
import pygame, sys

# ── What section is this? ──────────────
pygame.init()
screen = pygame.display.set_mode((800, 600))
clock  = pygame.time.Clock()
running = True

# ── What section is this? ──────────────
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
    screen.fill((0, 0, 0))
    pygame.display.flip()
    clock.tick(60)

# ── What section is this? ──────────────
pygame.quit()
sys.exit()
````

## 🎨 Section 3 — Colours in PyGame
PyGame represents every colour as a **tuple** of three numbers: `(Red, Green, Blue)`.
Each channel goes from `0` (none of that colour) to `255` (full intensity).

| Colour  | Tuple           |
|---------|----------------|
| Black   | `(0, 0, 0)`     |
| White   | `(255, 255, 255)`|
| Red     | `(255, 0, 0)`   |
| Green   | `(0, 255, 0)`   |
| Blue    | `(0, 0, 255)`   |
| Yellow  | `(255, 255, 0)` |
| Orange  | `(255, 165, 0)` |
| Purple  | `(128, 0, 128)` |

> **Tip:** Store colours as named variables at the top of your program so you never have to remember the numbers mid-code.


````python
# Run this to see the RGB mixer in action
# Try changing the numbers and re-running!

def describe_colour(r, g, b):
    """Print a mini colour description."""
    dominant = max((r, 'Red'), (g, 'Green'), (b, 'Blue'), key=lambda x: x[0])
    print(f'Colour ({r}, {g}, {b})')
    print(f'  Dominant channel : {dominant[1]}')
    print(f'  Brightness       : {(r + g + b) / (255 * 3) * 100:.0f}%')
    print()

describe_colour(255, 0, 0)     # pure red
describe_colour(255, 165, 0)   # orange
describe_colour(100, 100, 100) # grey
describe_colour(255, 255, 255) # white
````

## 📐 Section 4 — The Coordinate System
PyGame's coordinate system is **different from maths graphs**. The origin `(0, 0)` is in the **top-left** corner:

```
(0,0) ──────────────────────► x increases rightward
  │
  │
  ▼
  y increases downward
```

So for a window that is **800 wide × 600 tall**:
- Top-left  → `(0, 0)`
- Top-right → `(800, 0)`
- Bottom-left → `(0, 600)`
- Bottom-right → `(800, 600)`
- Centre → `(400, 300)`

> ⚠️ **Common mistake:** students expect `(0, 0)` to be bottom-left like in maths. In PyGame, moving *down* means *increasing* y.

## 🟦 Section 5 — Drawing Shapes
PyGame provides `pygame.draw` functions for common shapes. They all take the **surface** (your window) and a **colour** as the first two arguments.

| Shape     | Function                                          | Key arguments              |
|-----------|--------------------------------------------------|---------------------------|
| Rectangle | `pygame.draw.rect(surf, colour, (x, y, w, h))`  | x, y = top-left corner    |
| Circle    | `pygame.draw.circle(surf, colour, (cx, cy), r)` | cx, cy = centre; r = radius|
| Line      | `pygame.draw.line(surf, colour, p1, p2, width)` | p1, p2 = start/end points |

Add a **thickness** as the last argument to draw a hollow shape outline instead of filled:
```python
pygame.draw.rect(screen, RED, (50, 50, 200, 100), 4)   # 4-pixel border, not filled
```

## 🏃 Section 6 — Movement and Velocity
To move something, we store its **position** in variables and **change them** each frame.

The trick: use a **velocity** variable that says *how many pixels to move per frame*.

```python
# Position
x = 100
y = 200

# Velocity (speed + direction)
vel_x = 3   # move 3 pixels right each frame
vel_y = 2   # move 2 pixels down each frame

# Each frame in the update step:
x += vel_x   # new position = old position + velocity
y += vel_y
```

To **reverse direction**, negate the velocity:
```python
vel_x = -vel_x   # now moving left instead of right
```

## 🏓 Section 7 — Wall Bouncing
To make a ball bounce off walls, we check if it has reached an edge and reverse its velocity.

```
      ball_x - radius          ball_x + radius
           │                        │
   ◄───────┼────────  ball ─────────┼──────────►
           │                        │
       left edge                right edge
```

```python
# Hit the right wall?
if ball_x + radius >= WIDTH:
    vel_x = -vel_x   # reverse horizontal direction

# Hit the left wall?
if ball_x - radius <= 0:
    vel_x = -vel_x
```

> **Why use `ball_x + radius >= WIDTH` and not `ball_x >= WIDTH`?**  
> The ball's position is its *centre*. We need to account for its radius, otherwise half the ball disappears off-screen before bouncing.

## 📋 Quick Reference Card
Keep this handy for future PyGame projects.

### Essential functions
| Call | What it does |
|------|--------------|
| `pygame.init()` | Start all PyGame modules — always first |
| `pygame.display.set_mode((w, h))` | Create and return the game window |
| `pygame.display.set_caption("title")` | Set window title bar |
| `pygame.display.flip()` | Push completed frame to screen |
| `clock.tick(fps)` | Limit loop to `fps` frames per second |
| `pygame.event.get()` | Return all events since last frame |
| `screen.fill((r, g, b))` | Clear screen to a solid colour |
| `pygame.draw.circle(surf, col, (x,y), r)` | Filled circle |
| `pygame.draw.rect(surf, col, (x,y,w,h))` | Filled rectangle |
| `pygame.draw.line(surf, col, p1, p2, t)` | Line |
| `pygame.quit()` | Shut down PyGame cleanly |

### Event types to know
```python
pygame.QUIT      # user clicked the X button
pygame.KEYDOWN   # a key was just pressed
pygame.KEYUP     # a key was just released
```

### Common key constants

pygame.K_SPACE  pygame.K_RETURN  pygame.K_ESCAPE
