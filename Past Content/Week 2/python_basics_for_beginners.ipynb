## 2. 📦 Variables & Data Types

A **variable** is like a labelled box that stores a value. You can put something in the box, look at what's inside, or replace the contents.

You create a variable using `=` (the assignment operator):
```
box_label = value
```

Python has several built-in **data types** — the kinds of values a variable can hold:

| Type | Example | Description |
|------|---------|-------------|
| `str` | `"hello"` | Text (short for *string*) |
| `int` | `42` | Whole numbers |
| `float` | `3.14` | Decimal numbers |
| `bool` | `True` / `False` | True or False values |
| `list` | `[1, 2, 3]` | An ordered collection |
| `dict` | `{"key": "value"}` | A collection of key-value pairs |

````python
# Creating variables of different types
student_name = "Alex"          # str  (string — text)
student_age = 16               # int  (integer — whole number)
gpa = 3.75                     # float (decimal number)
is_enrolled = True             # bool (True or False)

# Print each variable
print(student_name)
print(student_age)
print(gpa)
print(is_enrolled)
````

````python
# You can check what type a variable is using type()
print(type(student_name))    # <class 'str'>
print(type(student_age))     # <class 'int'>
print(type(gpa))             # <class 'float'>
print(type(is_enrolled))     # <class 'bool'>
````

## 3. 🔤 Strings

A **string** is a sequence of characters (letters, numbers, symbols) wrapped in quotes. You can use single `'...'` or double `"..."` quotes — both work.

Strings have many useful built-in methods (actions you can perform on them).

````python
greeting = "Hello, World!"

# len() tells you how many characters are in the string
print(len(greeting))          # 13

# Convert to uppercase or lowercase
print(greeting.upper())       # HELLO, WORLD!
print(greeting.lower())       # hello, world!

# Replace part of a string
print(greeting.replace("World", "Python"))   # Hello, Python!

# Check if a string contains something
print("Hello" in greeting)    # True
print("Goodbye" in greeting)  # False
````

````python
# --- String Concatenation ---
# You can join (combine) strings together using +

first_name = "Ada"
last_name = "Lovelace"
full_name = first_name + " " + last_name
print(full_name)    # Ada Lovelace
````

````python
# --- f-strings (Formatted Strings) ---
# A cleaner way to insert variables into strings.
# Put f before the opening quote, then use {} to embed variables.

name = "Alex"
age = 16
print(f"My name is {name} and I am {age} years old.")

# You can even do maths inside the {}
print(f"Next year I will be {age + 1} years old.")
````

````python
# --- Indexing & Slicing ---
# Strings are like a list of characters. Each has a position (index).
# Indexing starts at 0, not 1!

word = "Python"
#        P y t h o n
#        0 1 2 3 4 5   <-- index positions

print(word[0])      # P  (first character)
print(word[3])      # h
print(word[-1])     # n  (last character — negative index counts from the end)

# Slicing: get a range of characters [start:stop]
# Note: stop index is NOT included
print(word[0:3])    # Pyt  (characters at index 0, 1, 2)
print(word[2:])     # thon (from index 2 to the end)
````

## 4. 🔢 Numbers & Arithmetic

Python can work like a calculator. There are two main number types:
- **int** — whole numbers: `5`, `-3`, `100`
- **float** — decimal numbers: `3.14`, `-0.5`, `2.0`

### Arithmetic Operators

| Operator | Symbol | Example | Result |
|----------|--------|---------|--------|
| Addition | `+` | `5 + 3` | `8` |
| Subtraction | `-` | `5 - 3` | `2` |
| Multiplication | `*` | `5 * 3` | `15` |
| Division | `/` | `5 / 2` | `2.5` |
| Floor Division | `//` | `5 // 2` | `2` (rounds down) |
| Modulo (remainder) | `%` | `5 % 2` | `1` |
| Exponent (power) | `**` | `5 ** 2` | `25` |

````python
a = 10
b = 3

print(a + b)    # 13
print(a - b)    # 7
print(a * b)    # 30
print(a / b)    # 3.333...
print(a // b)   # 3  (floor division — drops the decimal)
print(a % b)    # 1  (remainder: 10 = 3*3 + 1)
print(a ** b)   # 1000 (10 to the power of 3)
````

````python
# --- Useful built-in functions for numbers ---

print(abs(-7))          # 7    — absolute value (removes negative sign)
print(round(3.7))       # 4    — rounds to nearest whole number
print(round(3.14159, 2))# 3.14 — rounds to 2 decimal places
print(max(4, 9, 2))     # 9    — returns the largest value
print(min(4, 9, 2))     # 2    — returns the smallest value
````

## 5. 📋 Lists

A **list** stores multiple values in a single variable, in order. Lists are defined with square brackets `[]` and items are separated by commas.

Lists are **mutable** — you can change, add, and remove items after creating them.

````python
# Creating a list
fruits = ["apple", "banana", "cherry", "date"]
print(fruits)

# Accessing items by index (remember: starts at 0!)
print(fruits[0])     # apple
print(fruits[2])     # cherry
print(fruits[-1])    # date (last item)

# How many items are in the list?
print(len(fruits))   # 4
````

````python
# --- Modifying a list ---

numbers = [10, 20, 30, 40, 50]

# Change an item
numbers[1] = 99
print(numbers)       # [10, 99, 30, 40, 50]

# Add an item to the END
numbers.append(60)
print(numbers)       # [10, 99, 30, 40, 50, 60]

# Remove a specific item
numbers.remove(99)
print(numbers)       # [10, 30, 40, 50, 60]

# Remove the last item (and return it)
last = numbers.pop()
print(last)          # 60
print(numbers)       # [10, 30, 40, 50]
````

````python
# --- Useful list operations ---

scores = [85, 42, 97, 63, 78]

print(sum(scores))        # 365  — add all items
print(max(scores))        # 97   — highest value
print(min(scores))        # 42   — lowest value
print(sorted(scores))     # [42, 63, 78, 85, 97] — sorted copy (low to high)

# Check if something is in the list
print(97 in scores)       # True
print(100 in scores)      # False
````

## 6. 📖 Dictionaries

A **dictionary** stores data as **key-value pairs** — like a real dictionary where you look up a word (key) to find its definition (value).

Dictionaries are defined with curly braces `{}`.

```python
{"key": value, "key2": value2}
```

Keys are usually strings. Values can be anything.

````python
# A dictionary representing a student
student = {
    "name": "Jordan",
    "age": 17,
    "grade": "11B",
    "is_enrolled": True
}

# Access a value using its key
print(student["name"])        # Jordan
print(student["age"])         # 17

# A safer way to access — .get() won't crash if the key doesn't exist
print(student.get("grade"))       # 11B
print(student.get("score", "N/A")) # N/A (default if key is missing)
````

````python
# --- Modifying a dictionary ---

# Add a new key-value pair (or update existing)
student["score"] = 88
print(student)

# Delete a key-value pair
del student["is_enrolled"]
print(student)

# Useful dictionary methods
print(student.keys())     # all the keys
print(student.values())   # all the values
````

## 7. ✅ Booleans & Comparisons

A **Boolean** is a value that is either `True` or `False`. They're the result of **comparisons**.

### Comparison Operators

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `>=` | Greater than or equal | `5 >= 5` | `True` |
| `<=` | Less than or equal | `3 <= 5` | `True` |

### Logical Operators
You can combine comparisons with `and`, `or`, `not`.

````python
x = 10
y = 5

print(x == 10)    # True
print(x != y)     # True  (10 is not equal to 5)
print(x > y)      # True
print(x < y)      # False
````

````python
age = 17
has_permission = True

# 'and' — both conditions must be True
print(age >= 16 and has_permission)    # True

# 'or' — at least one condition must be True
print(age < 16 or has_permission)      # True

# 'not' — flips True to False and vice versa
print(not has_permission)              # False
````

## 8. 🔀 If / Elif / Else (Conditionals)

**Conditionals** let your program make decisions — run different code depending on whether something is True or False.

```python
if condition:
    # code runs if condition is True
elif another_condition:
    # code runs if the first was False, but this is True
else:
    # code runs if ALL above conditions were False
```

> ⚠️ **Indentation is critical in Python!** Code inside an `if` block must be indented (4 spaces or 1 tab). Python uses indentation to know what belongs where.

````python
temperature = 25

if temperature > 30:
    print("It's hot! Wear sunscreen.")
elif temperature > 20:
    print("Nice weather! A t-shirt is fine.")
elif temperature > 10:
    print("A bit chilly. Grab a jacket.")
else:
    print("It's cold! Bundle up.")
````

````python
# Another example — checking if a number is even or odd
# Hint: A number is even if it has no remainder when divided by 2 (n % 2 == 0)

number = 7

if number % 2 == 0:
    print(f"{number} is even")
else:
    print(f"{number} is odd")
````

## 9. 🔄 For Loops

A **for loop** lets you repeat an action for each item in a sequence (like a list or a range of numbers). Instead of writing the same code 10 times, you write it once and let the loop handle it.

```python
for item in collection:
    # do something with item
```

````python
# Looping over a list
colours = ["red", "green", "blue", "yellow"]

for colour in colours:
    print(f"I like the colour {colour}")
````

````python
# --- range() ---
# range(n) generates numbers from 0 up to (but not including) n

for i in range(5):
    print(i)          # 0, 1, 2, 3, 4

print("---")

# range(start, stop)
for i in range(1, 6):
    print(i)          # 1, 2, 3, 4, 5

print("---")

# range(start, stop, step) — count by 2s
for i in range(0, 11, 2):
    print(i)          # 0, 2, 4, 6, 8, 10
````

````python
# --- Useful: building a total with a loop ---

prices = [4.99, 12.50, 3.25, 8.00, 6.75]
total = 0

for price in prices:
    total = total + price    # can also write: total += price

print(f"Total cost: ${total:.2f}")
````

## 10. 🔁 While Loops

A **while loop** keeps running as long as a condition is `True`. It's useful when you don't know ahead of time how many times you need to loop.

```python
while condition:
    # keep doing this
```

> ⚠️ **Warning: Infinite Loops!** If the condition never becomes `False`, the loop runs forever and freezes your program. Always make sure something inside the loop will eventually make the condition `False`.

````python
# Count from 1 to 5 using a while loop
count = 1

while count <= 5:
    print(count)
    count += 1    # IMPORTANT: increase count or this runs forever!

print("Done!")
````

````python
# --- break and continue ---

# 'break' exits the loop immediately
# 'continue' skips to the next iteration

# Example: find the first number divisible by 7 between 1 and 50
num = 1
while num <= 50:
    if num % 7 == 0:
        print(f"First multiple of 7: {num}")
        break          # stop the loop once we find it
    num += 1
````

````python
# Example using continue — skip even numbers and only print odds
for n in range(1, 11):
    if n % 2 == 0:
        continue    # skip even numbers
    print(n)
````

## 11. 🛠️ Functions

A **function** is a reusable block of code that performs a specific task. Instead of copy-pasting the same code in multiple places, you write it once as a function and *call* it whenever you need it.

```python
def function_name(parameter1, parameter2):
    # code goes here
    return result
```

- `def` — tells Python you're defining a function
- **Parameters** — inputs the function accepts
- `return` — sends a value back to whoever called the function

````python
# A simple function with no parameters
def say_hello():
    print("Hello there!")

# Call the function (run it) using its name followed by ()
say_hello()
say_hello()
````

````python
# A function with a parameter
def greet(name):
    print(f"Hello, {name}! Nice to meet you.")

greet("Alice")
greet("Bob")
````

````python
# A function that takes parameters AND returns a value
def add(a, b):
    result = a + b
    return result

total = add(10, 5)
print(total)            # 15
print(add(100, 200))    # 300
````

````python
# Default parameter values — used if no argument is given
def power(base, exponent=2):
    return base ** exponent

print(power(3))       # 9   (3 squared — uses default exponent of 2)
print(power(3, 3))    # 27  (3 cubed)
print(power(2, 10))   # 1024
````

````python
# A more realistic function — calculate average of a list
def calculate_average(numbers):
    if len(numbers) == 0:
        return 0    # avoid dividing by zero!
    total = sum(numbers)
    average = total / len(numbers)
    return average

my_scores = [85, 92, 78, 95, 88]
avg = calculate_average(my_scores)
print(f"Average score: {avg}")
````

## 12. 🏁 Putting It All Together

Let's combine everything we've learned into a mini project — a **Student Grade Report** generator.

This program will:
- Store student data in a dictionary
- Use a function to calculate their grade
- Use a loop to process multiple students
- Use conditionals to assign letter grades
- Print a formatted report

````python
# --- Student Grade Report ---

# A list of student dictionaries (each dict is one student)
students = [
    {"name": "Alice",   "scores": [88, 92, 95, 87, 91]},
    {"name": "Bob",     "scores": [70, 65, 72, 68, 74]},
    {"name": "Charlie", "scores": [55, 60, 48, 52, 58]},
    {"name": "Diana",   "scores": [99, 97, 100, 95, 98]},
]

# Function to calculate the average of a list of numbers
def get_average(scores):
    return sum(scores) / len(scores)

# Function to return a letter grade based on average score
def get_letter_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F"

# Print the report
print("=" * 40)
print("       STUDENT GRADE REPORT")
print("=" * 40)

for student in students:
    name = student["name"]
    scores = student["scores"]
    average = get_average(scores)
    grade = get_letter_grade(average)
    
    print(f"Student: {name}")
    print(f"  Scores:  {scores}")
    print(f"  Average: {average:.1f}")
    print(f"  Grade:   {grade}")
    print("-" * 40)

print("End of Report")
````
